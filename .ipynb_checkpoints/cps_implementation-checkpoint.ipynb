```python
# CPS Anomaly Detection – Implementation & Testing Notebook
# This notebook implements and tests the anomaly-detection model
# described in the design & planning phases of the capstone project.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.ensemble import IsolationForest
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# For reproducibility
RANDOM_STATE = 42

pd.set_option("display.max_columns", 50)
```

```python
# Data loading step
# We use the same two CPS-style datasets as in the Streamlit app.

IBM_URL = "https://raw.githubusercontent.com/IBM/iot-predictive-analytics/master/data/iot_sensor_dataset.csv"
NAB_URL = "https://raw.githubusercontent.com/numenta/NAB/master/data/realKnownCause/ambient_temperature_system_failure.csv"

print("Loading IBM IoT dataset...")
ibm_df = pd.read_csv(IBM_URL)

print("Loading NAB ambient temperature dataset...")
nab_df = pd.read_csv(NAB_URL)

print("IBM shape:", ibm_df.shape)
print("NAB shape:", nab_df.shape)

ibm_df.head()
```

```python
# Basic preprocessing for the IBM CPS dataset

# 1. Sanitize column names (similar to app.py)
ibm_df = ibm_df.copy()
ibm_df.columns = [c.strip().replace(" ", "_").replace("/", "_") for c in ibm_df.columns]

# 2. Identify numeric columns
numeric_cols = ibm_df.select_dtypes(include=[np.number]).columns.tolist()
print("Numeric columns:", numeric_cols)

# 3. Keep only numeric features for modeling
X_raw = ibm_df[numeric_cols].copy()

# 4. Check missing values
missing_counts = X_raw.isna().sum()
print("Missing values per numeric column:")
print(missing_counts)

# 5. Simple missing-value handling: mean imputation
X = X_raw.fillna(X_raw.mean())

# 6. Standard scaling (important for IsolationForest & PCA)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Scaled feature matrix shape:", X_scaled.shape)
```

```python
# IsolationForest model implementation (unsupervised anomaly detection)

iso_model = IsolationForest(
    n_estimators=200,
    contamination=0.01,  # assume ~1% anomalies
    random_state=RANDOM_STATE,
)

iso_model.fit(X_scaled)

# anomaly scores: higher = more anomalous
iso_scores = -iso_model.score_samples(X_scaled)

# add scores back to a dataframe
results_df = ibm_df.copy()
results_df["iso_anomaly_score"] = iso_scores

# rank and flag top anomalies (top 1%)
results_df["iso_rank"] = results_df["iso_anomaly_score"].rank(
    method="first", ascending=False
)
threshold_rank = int(len(results_df) * 0.01)  # top 1%
results_df["iso_is_anomaly"] = results_df["iso_rank"] <= threshold_rank

results_df[["iso_anomaly_score", "iso_is_anomaly"]].head()
```

```python
# Initial tests: basic sanity checks on the IsolationForest output

n_total = len(results_df)
n_anom = results_df["iso_is_anomaly"].sum()

print(f"Total records: {n_total}")
print(f"Flagged anomalies (top 1% by score): {n_anom} ({n_anom / n_total:.2%})")

print("\nAnomaly score summary:")
print(results_df["iso_anomaly_score"].describe())

# Plot distribution of anomaly scores
plt.figure(figsize=(8, 4))
sns.histplot(results_df["iso_anomaly_score"], bins=50, kde=True)
plt.title("IsolationForest anomaly score distribution")
plt.xlabel("Anomaly score")
plt.ylabel("Count")
plt.show()
```

```python
# PCA reconstruction error model (proxy for autoencoder)

pca = PCA(n_components=0.95, svd_solver="full", random_state=RANDOM_STATE)
pca.fit(X_scaled)

X_trans = pca.transform(X_scaled)
X_recon = pca.inverse_transform(X_trans)

recon_errors = np.mean((X_scaled - X_recon) ** 2, axis=1)

results_df["pca_recon_error"] = recon_errors

# Flag top 1% highest reconstruction error as anomalies
results_df["pca_rank"] = results_df["pca_recon_error"].rank(
    method="first", ascending=False
)
results_df["pca_is_anomaly"] = results_df["pca_rank"] <= threshold_rank

print("PCA reconstruction error summary:")
print(results_df["pca_recon_error"].describe())

plt.figure(figsize=(8, 4))
sns.histplot(results_df["pca_recon_error"], bins=50, kde=True)
plt.title("PCA reconstruction error distribution")
plt.xlabel("Reconstruction error")
plt.ylabel("Count")
plt.show()
```

```python
# Compare how many points each model flags as anomalies

print("IsolationForest anomalies:", results_df["iso_is_anomaly"].sum())
print("PCA anomalies:", results_df["pca_is_anomaly"].sum())

# How much do they overlap?
overlap = (results_df["iso_is_anomaly"] & results_df["pca_is_anomaly"]).sum()
print("Overlap (both models flag anomaly):", overlap)

# Show a small sample of anomalies for inspection
results_df[results_df["iso_is_anomaly"] | results_df["pca_is_anomaly"]].head(10)
```

```python
# Example of robust data-loading with error handling (for your report)

def safe_load_csv(url: str) -> pd.DataFrame:
    """
    Load a CSV file from a URL with simple error handling.
    This mirrors the defensive coding approach used in the Streamlit app.
    """
    try:
        if not url.lower().startswith(("http://", "https://")):
            raise ValueError("URL must start with http:// or https://")
        df = pd.read_csv(url)
        if df.empty:
            raise ValueError("Loaded dataframe is empty")
        return df
    except Exception as e:
        print(f"[ERROR] Could not load dataset from {url}: {e}")
        return pd.DataFrame()

# Test the helper with an intentionally bad URL
bad_df = safe_load_csv("not_a_url")
print("Bad DF shape (should be 0 rows):", bad_df.shape)
```

### Data Security and Compliance Notes

- Only **public CPS-style datasets** from GitHub are used (no PII or PHI).
- Data is downloaded inside a local virtual environment (`C:\cps_app\venv`) and is not shared externally.
- No raw data is logged to external services; only derived metrics and plots are saved in the notebook.
- If extended to real CPS logs, the same code would be wrapped with:
  - OS-level access controls on the data folder.
  - Encrypted storage (e.g., BitLocker or an encrypted volume).
  - Version control (Git/GitHub) for code only, not for sensitive raw data.
